# 01 — Data exploration

Inspect synthetic EEG (or PhysioNet arrays after preprocessing) and class balance.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))

import numpy as np
import matplotlib.pyplot as plt
from data.synthetic import generate_dataset
from src.config import CLASS_NAMES, SAMPLING_RATE

X, y = generate_dataset(n_per_class=20)
print(X.shape, y.shape, {CLASS_NAMES[i]: int((y==i).sum()) for i in range(len(CLASS_NAMES))})
t = np.arange(X.shape[-1]) / SAMPLING_RATE
plt.figure(figsize=(8, 3))
plt.plot(t, X[0, 8], label='C3')
plt.plot(t, X[0, 12], label='C4')
plt.legend(); plt.xlabel('Time (s)'); plt.title(CLASS_NAMES[int(y[0])])
plt.show()
```
